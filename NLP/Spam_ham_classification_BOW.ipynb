# **BOW, TF-IDF & Machine Learning**

1. Preprocessing and cleaning
2. Train-test-Split
3. BOW & TF-IDF(sentences ----> vectors)
4. Train our models

```python
import pandas as pd
messages=pd.read_csv('/content/SMSSpamCollection.txt',sep='\t',names=['label','message'])
```

```python
messages
```

```python
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
```

```python
ps=PorterStemmer()
```

```python
corpus=[]
for i in range(0,len(messages)):
  review=re.sub('[^a-zA-Z]', ' ', messages['message'][i])
  review=review.lower()
  review=review.split()
  review=[ps.stem(word)for word in review if not word in stopwords.words('english')]
  review=' '.join(review)
  corpus.append(review)
```

```python
corpus
```

# **Train Test Split**

```python
## Output features
y=pd.get_dummies(messages['label'])
```

```python
y=y.iloc[:,0].values
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(corpus, y, test_size=0.20)
```

```python
X_train
```

```python
print(len(X_train))
print(len(y_train))
```

# **Create Bag of Words**

```python
### Create the bag of words model with ngram
from sklearn.feature_extraction.text import CountVectorizer
## Binary Bag of words
cv=CountVectorizer(max_features=2500, binary=True, ngram_range=(1,2))
```

```python
## Independent features
X_train=cv.fit_transform(X_train).toarray()
X_test=cv.transform(X_test).toarray()
```

```python
cv.vocabulary_
```

# **Machine Learning model for prediction**

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
spam_detect_model=MultinomialNB().fit(X_train,y_train)
```

```python
spam_detect_model.predict(X_test)
```

# **Performance Metrics**

```python
from sklearn.metrics import accuracy_score, classification_report
```

```python
accuracy_score(y_test,spam_detect_model.predict(X_test))
```

```python
print(classification_report(y_test,spam_detect_model.predict(X_test)))
```
