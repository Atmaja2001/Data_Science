```python
import pandas as pd
messages=pd.read_csv('/content/SMSSpamCollection.txt',sep='\t',names=['label','message'])
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
ls = WordNetLemmatizer()
```

```python
import re
import nltk
nltk.download('stopwords')
```

```python
import nltk
nltk.download('wordnet')
```

```python
corpus=[]
for i in range(0,len(messages)):
  review=re.sub('[^a-zA-Z]', ' ', messages['message'][i])
  review=review.lower()
  review=review.split()
  review=[ls.lemmatize(word)for word in review ]
  review=' '.join(review)
  corpus.append(review)
```

The list comprehension [[i, j, k] for i,j,k in zip(list(map(len,corpus)), corpus, messages['message']) if i < 1] produces a list of lists. Each inner list contains three elements:

1. i: The length of the preprocessed message (j).
2. j: The preprocessed message from the corpus list.
3. k: The original message from the messages['message'] column.

```python
[[i, j, k] for i,j,k in zip(list(map(len,corpus)), corpus, messages['message']) if i < 1]
```

```python
from gensim.utils import simple_preprocess
from nltk.text import sent_tokenize
```

```python
words=[]
for sent in corpus:
  sent_token=sent_tokenize(sent)
  for sent in sent_token:
    words.append(simple_preprocess(sent))
```

# **Create a AvgWord2Vec**

```python
!pip install tqdm
```

```python
from tqdm import tqdm
```

```python
import numpy as np
def avg_word2vec(doc):
  # remove out-of-vocabulary words
  # Check if there are any words from doc present in model's vocabulary
  words_in_vocab = [word for word in doc if word in model.wv.index_to_key]
  if words_in_vocab:
    return np.mean([model.wv[word] for word in words_in_vocab], axis=0)
  else:
    # Return a zero vector or handle as appropriate for words not in vocabulary
    return np.zeros(model.vector_size)

#apply for the entire sentences
X = []
for i in tqdm(range(len(words))):
  X.append(avg_word2vec(words[i]))
```
