# **About Dataset**
**Context This is a small subset of dataset of Book reviews from Amazon Kindle Store category.**

Content 5-core dataset of product reviews from Amazon Kindle Store category from May 1996 - July 2014. Contains total of 982619 entries. Each reviewer has at least 5 reviews and each product has at least 5 reviews in this dataset. Columns

asin - ID of the product, like B000FA64PK
helpful - helpfulness rating of the review - example: 2/3.
overall - rating of the product.
reviewText - text of the review (heading).
reviewTime - time of the review (raw).
reviewerID - ID of the reviewer, like A3SPTOKDG7WBLN
reviewerName - name of the reviewer.
summary - summary of the review (description).
unixReviewTime - unix timestamp.
Acknowledgements This dataset is taken from Amazon product data, Julian McAuley, UCSD website. http://jmcauley.ucsd.edu/data/amazon/

License to the data files belong to them.

**Preprocessing & cleaning**

```python
## positive review as 1 and negative review as no
data['rating']=data['rating'].apply(lambda x : 0 if x <3 else 1)
```

```python
data['rating']
```

```python
data['rating'].value_counts()
```

```python
## 1. Lower all the cases
data['reviewText']=data['reviewText'].str.lower()
```

```python
import re
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
```

```python
from bs4 import BeautifulSoup
```

```python
## 2. removing special characters
data['reviewText']=data['reviewText'].apply(lambda x:re.sub(r'[^a-z A-Z 0-9]+', '', x))
## removing the stopwords
data['reviewText']=data['reviewText'].apply(lambda x: " ".join([y for y in x.split() if y not in stopwords.words('english') ]))
## removing urls
data['reviewText']=data['reviewText'].apply(lambda x: re.sub(r'(http|https|ftp|ssh)://([\w_-]+(?:(?:\.[\w_-]+)+))([\w.,@?^=%&:/~+#-]*[\w@?^=%&/~+#-])?', '' , str(x)))

##remove html tags
data['reviewText']=data['reviewText'].apply(lambda x: BeautifulSoup(x, 'lxml').get_text())
## Remove any additional spaces
data['reviewText']=data['reviewText'].apply(lambda x: " ".join(x.split()))
```

```python
data.head()
```

```python
## Lemmatize the word into lemma
from nltk.stem import WordNetLemmatizer
```

```python
lemmtizer=WordNetLemmatizer()
```

```python
def lemmatize_words(text):
  return " ".join([lemmtizer.lemmatize(word) for word in text.split()])
```

```python
import nltk
nltk.download('wordnet')
```

```python
data['reviewText']=data['reviewText'].apply(lambda text: lemmatize_words(text))
```

```python
data.head()
```

```python
## Train_test_split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(data['reviewText'], data['rating'], test_size=0.20)
```

**BOW embedding**

```python
from sklearn.feature_extraction.text import CountVectorizer
bow=CountVectorizer()
X_train_bow=bow.fit_transform(X_train).toarray()
X_test_bow=bow.transform(X_test).toarray()
```

**TF-TDF embedding**

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer()
X_train_tfidf=tfidf.fit_transform(X_train).toarray()
X_test_tfidf=tfidf.transform(X_test).toarray()
```

**Word2Vec embedding**

```python
import gensim
from gensim.models import Word2Vec, KeyedVectors
```

```python
tokenized_reviews_train = X_train.apply(lambda x: x.split())
tokenized_reviews_test = X_test.apply(lambda x: x.split())
```

```python
word2vec_model = gensim.models.Word2Vec(
    sentences=tokenized_reviews_train,
    vector_size=100,
    window=5,
    min_count=1,
    sg=0
)
```

```python
def get_word2vec_embeddings(text_list, model):
    vectors = []
    for word in text_list:
        if word in model.wv:
            vectors.append(model.wv[word])
    if vectors:
        return sum(vectors) / len(vectors)
    else:
        return [0.0] * model.vector_size
```

```python
X_train_word2vec = tokenized_reviews_train.apply(lambda x: get_word2vec_embeddings(x, word2vec_model))
X_test_word2vec = tokenized_reviews_test.apply(lambda x: get_word2vec_embeddings(x, word2vec_model))
```

```python
X_train_word2vec = pd.DataFrame(X_train_word2vec.tolist())
X_test_word2vec = pd.DataFrame(X_test_word2vec.tolist())
```

```python
print('X_train_word2vec shape:', X_train_word2vec.shape)
print('X_test_word2vec shape:', X_test_word2vec.shape)

display(X_train_word2vec.head())
display(X_test_word2vec.head())
```

# **NaiveBayes model**

```python
from sklearn.naive_bayes import GaussianNB
nb_model_bow=GaussianNB().fit(X_train_bow,y_train)
nb_model_tfidf=GaussianNB().fit(X_train_tfidf,y_train)
```

```python
y_pred_bow=nb_model_bow.predict(X_test_bow)
```

```python
y_pred_tfidf=nb_model_tfidf.predict(X_test_tfidf)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
print("BOW accuracy",accuracy_score(y_test, y_pred_bow))
```

```python
print("TF-TDF accuracy",accuracy_score(y_test, y_pred_tfidf))
```
